# Text Preprocessing Pipeline

Before converting text into numbers, we must **clean and standardize** text.

This process is called:

# Text Preprocessing Pipeline

---

# Why Preprocessing is Required

Consider Raw Sentences:

"I love Machine Learning!!!"

"I LOVE machine learning"

"I love ML"

For humans → Same meaning  
For machines → Completely different

This creates:

- Vocabulary explosion
- Poor learning
- Data inconsistency

---

# NLP Preprocessing Pipeline

Raw Text → Tokenization → Lowercase → Stopword Removal → Stemming → Lemmatization

---

# Step 1 — Tokenization

Tokenization converts sentence into words

Example:

"I love NLP"

↓

["I", "love", "NLP"]

Mathematically:

Sentence:

$S = [w_1, w_2, w_3, ..., w_n]$

---

# Step 2 — Lower Casing

"I Love NLP"

↓

"i love nlp"

Why?

"I" and "i" should represent same word

---

# Step 3 — Stop Words Removal

Common words:

- is
- the
- and
- are
- of

Example:

"This is a machine learning model"

↓

["machine", "learning", "model"]

---

# Step 4 — Stemming

Reducing word to root

Example:

running → run  
played → play  
learning → learn

But:

Sometimes incorrect

Example:

studies → studi

---

# Step 5 — Lemmatization

Better version of stemming

running → run  
better → good  
studies → study

Lemmatization understands **word meaning**

---

# Final Pipeline Example

Raw Sentence:

"I am learning machine learning and loving it"

After Processing:

["learn", "machine", "learn", "love"]

---

# Why This Matters

After preprocessing:

- Vocabulary size reduces
- Model learns better
- Computation reduces
- Performance improves

---

# Bridge to Next Step

Now we have clean tokens:

["learn", "machine", "learn"]

Next step:

Convert words into numbers

```python
# Text Preprocessing Example

import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet

# Downloads (Run once)
#nltk.download('averaged_perceptron_tagger_eng')
#nltk.download('punkt')
#nltk.download('stopwords')
#nltk.download('wordnet')
#nltk.download('averaged_perceptron_tagger')

sentence = "I would perferably be learning machine learning and enjoy it"

# Lowercase
sentence = sentence.lower()

# Tokenization
tokens = nltk.word_tokenize(sentence)

# Remove punctuation
tokens = [re.sub(r'[^\w\s]', '', word) for word in tokens]

# Remove empty strings
tokens = [word for word in tokens if word != ""]

# Remove stopwords
stop_words = set(stopwords.words('english'))
tokens = [word for word in tokens if word not in stop_words]

# Stemming
#stemmer = PorterStemmer()
#stemmed = [stemmer.stem(word) for word in tokens]

# Function for Lemmatization with POS tagging

def get_wordnet_pos(word):
    tag = pos_tag([word])[0][1][0].upper()
    tag_dict = {
        "J": wordnet.ADJ,
        "N": wordnet.NOUN,
        "V": wordnet.VERB,
        "R": wordnet.ADV
    }
    return tag_dict.get(tag, wordnet.NOUN)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(word, get_wordnet_pos(word)) for word in tokens]

print("Original Sentence:")
print("I would perferably be learning machine learning and enjoy it")

print("\nTokens:")
print(tokens)

#print("\nStemmed:")
#print(stemmed)

print("\nLemmatized:")
print(lemmatized)
```

Observation:

After preprocessing:

- Words become standardized
- Vocabulary reduces
- Learning becomes easier

Example:

learning → learn  
loving → love  

This reduces dimensionality of vocabulary

# Vocabulary Problem

After preprocessing, we obtain **tokens (words)**:

Example:

"I love NLP"

↓

["i", "love", "nlp"]

But neural networks **cannot understand words**.

They only understand **numbers**.

So we must convert words → numbers

---

# Vocabulary Creation

Step 1: Collect all unique words

Example Dataset:

"I love NLP"  
"I love Deep Learning"  
"Deep learning is powerful"

Vocabulary:

V = {i, love, nlp, deep, learning, is, powerful}

---

# Word Indexing

Assign index to each word:

i → 0  
love → 1  
nlp → 2  
deep → 3  
learning → 4  
is → 5  
powerful → 6

This is called:

# Word Dictionary Mapping

---

# Convert Sentence to Numbers

"I love NLP"

↓

[0, 1, 2]

"Deep learning is powerful"

↓

[3, 4, 5, 6]

---

# Mathematical Representation

Vocabulary:

$V = \{w_1, w_2, w_3, ... , w_n\}$

Index mapping:

$f(w_i) = i$

Sentence:

$S = [w_1, w_2, w_3]$

Numeric Representation:

$S = [1,2,3]$

---

# Why This is Important

Now we converted:

Text → Numbers

But there is still a problem:

[0,1,2] does NOT capture meaning

Example:

king → 1  
queen → 2  

Does this mean queen > king?

No.

Numbers introduce **false relationships**

This leads to:

One Hot Encoding (Next Section)

```python
# Vocabulary Creation Example

sentences = [
    "i love nlp",
    "i love deep learning",
    "deep learning is powerful"
]

# Tokenization
tokens = [sentence.split() for sentence in sentences]

# Create vocabulary
vocab = set()

for sentence in tokens:
    for word in sentence:
        vocab.add(word)

# Convert to dictionary
word2index = {word: i for i, word in enumerate(vocab)}

print("Vocabulary:\n")
print(word2index)

# Convert sentence to numeric

example = "i love nlp".split()
numeric = [word2index[word] for word in example]

print("\nSentence:")
print(example)

print("\nNumeric Representation:")
print(numeric)
```

Observation:

- Words converted to numbers
- Vocabulary created
- Sentence converted to numeric sequence

But numbers do not represent meaning

Example:

king → 1  
queen → 2  

Model assumes numeric relationship

This is incorrect

We need better representation

# One Hot Encoding

In Vocabulary Mapping:

king → 0  
queen → 1  
man → 2  
woman → 3  

Sentence:

"I love NLP"

↓

[0,1,2]

But numeric indexing introduces **false relationships**

queen (1) > king (0)

This is incorrect.

---

# Solution — One Hot Encoding

Instead of using numbers directly,  
we represent each word as **vector**

Example Vocabulary:

{king, queen, man, woman}

Vector Representation:

king  → [1,0,0,0]  
queen → [0,1,0,0]  
man   → [0,0,1,0]  
woman → [0,0,0,1]

Each word has **unique vector**

---

# Properties of One Hot Encoding

- Only one position is 1
- All others are 0
- Vector size = Vocabulary size

---

# Mathematical Representation

Vocabulary size = V

Each word represented as:

One Hot Vector:

Dimension = V

---

# Example

Vocabulary:

{I, love, NLP}

I → [1,0,0]  
love → [0,1,0]  
NLP → [0,0,1]

Sentence:

"I love NLP"

↓

[[1,0,0],  
 [0,1,0],  
 [0,0,1]]

---

# Why One Hot Encoding is Better

- No numeric bias
- No false ordering
- Unique representation

But still problem exists:

king → [1,0,0,0]  
queen → [0,1,0,0]

Similarity = 0

But king and queen are related.

One Hot Encoding cannot capture similarity

This leads to:

Word Embeddings
